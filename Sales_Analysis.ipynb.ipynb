```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

print(region_sales)
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(5)

print(top_products)
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

category_sales.plot(kind='bar', figsize=(8,5))

plt.title("Category Wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Month'] = df['Order Date'].dt.month

monthly_sales = df.groupby('Month')['Sales'].sum()

monthly_sales.plot(kind='line', marker='o', figsize=(10,5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
category_profit = df.groupby('Category')['Profit'].sum()

print(category_profit)
```

```python
region_profit = df.groupby('Region')['Profit'].sum()

print(region_profit)
```

```python
top_customers = (
    df.groupby('Customer Name')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_customers)
```

```python
plt.figure(figsize=(9,5))

plt.scatter(df['Sales'], df['Profit'])

plt.title("Sales vs Profit Analysis")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.scatterplot(data=df, x='Discount', y='Profit')

plt.title("Discount vs Profit")
plt.show()
```

```python
category_sales = df.groupby('Category')['Sales'].sum()

print("Highest Sales Category:")
print(category_sales.idxmax())
```

```python
region_profit = df.groupby('Region')['Profit'].sum()

print("Highest Profit Region:")
print(region_profit.idxmax())
```

```python
customer_sales = df.groupby('Customer Name')['Sales'].sum()

print("Top Customer:")
print(customer_sales.idxmax())
```

```python
category_profit = df.groupby('Category')['Profit'].sum()

print("Lowest Profit Category:")
print(category_profit.idxmin())
```

```python
profit_margin = (df['Profit'].sum() / df['Sales'].sum()) * 100

print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
df.to_csv("cleaned_sales_data.csv", index=False)

print("Cleaned Dataset Saved Successfully!")
```
